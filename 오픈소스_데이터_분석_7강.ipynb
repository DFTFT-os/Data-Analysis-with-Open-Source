(그 1시그마 2시그마 95% 3시그마 99%)

- 제거: 결측치나 이상치를 포함하는 행/열 삭제 dropna() 신중하게 써야함
- 대치: 결측치나 이상치를 적절한 값으로 대체, 일반적으로 많이 쓰임 (하나의 다른 정해진 값으로 대체하는거에용)

### 7-1 데이터 결측치 찾기

```python
import pandas as pd
import numpy as np

data = {'이름': ['김철수', '이영희', '박민수', '최지훈', '정소희'],
        '나이': [25, 30, np.nan, 22, 35],
        '도시': ['서울', None, '인천', '서울', '대전'],
        '점수': [90, 85, np.nan, 80, 92]}

df = pd.DataFrame(data)

## 결측치 여부 확인
df.isnull()
df.isna()

## 열별, 행별 결측치 개수 확인
print(df)
df.isnull().sum()
df.isnull().sum(axis=1) #방향 바꿔줌 가로방향으로 묶어주세요

## 특정 열, 행 결측치 확인
df[df.isnull().any(axis=1)]
df[df['나이'].isnull()]

## 결측치가 아닌 항목 확인
df.notnull()

## 결측치 비율 (데이터 품질 평가할 때 사용)
df.isnull().sum().sum() / (df.shape[0] * df.shape[1]) * 100
#전체 데이터 항목 중에 null 이 몇개일까용? 전체개수는 가로크기 * 세로크기 앞에선 세로 각각 결측치 합한다음 그걸 전체 합쳐죠요 그러면 15% 결측치가 있다는게 나와용
```
