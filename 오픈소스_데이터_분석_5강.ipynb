## 5-1 CSV 형식 저장

ㅇ 데이터 저장
1. 수집/저장
2. 분석한 데이터 저장

ㅇ 데이터 저장할 때 쓰는 frame = pandas

ㅇ 데이터 저장: 수집한 데이터를 영구적으로 사용할 수 있도록 적절하게 관리하는 과정
방식: file, database
- 운영체제에서 제공하는 파일 시스템 상에서 데이터를 독립적인 단위로 저장
- 텍스트, csv, json 등 다양한 형태 가능
- 호환성이 뛰어나고(대부분의 프로그램, 소프트웨어 등등이 지원함) 여러 환경에서 파일/읽기 쓰기 가능
- 대규모 데이터 처리에 비효율적 (순차적으로 접근해야하기 때문) 데이터 무결성과 보안 측면에서 취약

ㅇCSV; Comma Separated Version
- dataframe은 csv와 구조적인 유사성이 있음


```python
import pandas as pd

data = {
    'student_id': [101, 102, 103, 104, 105],
    'database_score': [85, 76, 92, 63, 88],
    'cloudcomputing_score': [78, 82, 95, 70, 84],
    'python_score': [92, 78, 85, 75, 91],
    'watch_rate': [0.95, 0.87, 0.99, 0.80, 0.93]
}
## DataFrame 생성
df = pd.DataFrame(data)
df

## CSV 형식 저장
df.to_csv('students_analysis.csv', encoding='utf-8', index=False)
```

## 5-2 JSON 형식 저장

ㅇ JSON; JavaScript Object Notation
- 계층 구조를 가진 데이터를 저장하는데 유용
- json 또는 pandas 라이브러리를 사용하여 데이터를 JSON 파일로 저장 가능
아래 형태는 csv로 만들긴 곤란함 관심사는 3개니까..
data = {
    "이름": "홍길동",
    "나이": 25,
    "거주지": "서울",
    "관심사": ["프로그래밍", "데이터 분석", "여행"]
}

```python
### 라이브러리 임포트
import json
import pandas as pd

data = {
    "이름": "홍길동",
    "나이": 25,
    "거주지": "서울",
    "관심사": ["프로그래밍", "데이터 분석", "여행"]
}

## json.dump를 이용한 저장
with open('output.json', mode='w', encoding='utf-8') as f:
    json.dump(data, f, indent=4, ensure_ascii=False)
    #indent 들여쓰기

## DataFrame을 이용한 저장
df = pd.DataFrame([data])
df.to_json('out_df.json', orient='records', indent=4, force_ascii=False)
#관심사가 3개라 3개로 표현됨 []리스트로 묶어주면 하나로 나옴
#orient
```
