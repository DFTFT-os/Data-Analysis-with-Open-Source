# 오픈소스 기반 데이터 분석 6강 - 데이터 전처리 1

데이터전처리 어렵고 규모가 커서 2번에 나눠서 해볼게요오옹

ㅇ 데이터 전처리의 필요성
- GIGO; Garbage In, Garbage Out
- 잘못되거나 부정확한 데이터 입력하면 어떤 정교한 분석 기법을 적용하더라도 잘못된 결과가 출력된다
- 데이터 품질 향상을 통해 분석 결과의 질을 향상시키기 위한 방안이 요구
- 데이터 규모 증가, 신기술 대중화로 다양하고 새로운 유형의 데이터 수집, 여러 이질적인 소스로부터 데이터 수집

ㅇ 전처리 대상 데이터
- 데잍터 수집되지 않음: 결측치, 누락된 값
- 일관성이 없는 데이터: mouse, 마우스
- 발생할 수 없는 값(부정확한 값): 판매 수량인데 -10

ㅇ 데이터 전처리의 정의
- 데이터 분석에 앞서 원시 데이터 raw data 를 정제하고 변환하여 분석에 적합한 형태로 만드는 일련의 작업
1. 측정 (구조 이해, 필요 여부 판단)
2. 정제 (정합석 확보, 오류 수정)
3. 통합 (데이터셋 결합)
4. 축소 (크기 축소, 샘플링, 특징 추출)
5. 변환 (형태 변환, 정규화, 요약)
> 5단계 순환, 지속/반복적 수행

ㅇ 어려운 이유
- 방대한 양, 수치형 데이터 뿐만 아니라 텍스트, 이미지, 동영상, 로그 파일 등 다양한 형식이 존재
- 주관적 판단의 개입 여지, 정해진 정답 없고 상황에 따라 적절한 결정
- 다양한 처리 방법, 처리 방법 고정 X, 다양한 접근법 존재
- 시간과 노력 소모, 여러 단계를 반복적으로 확인하고 검토하는 과정
