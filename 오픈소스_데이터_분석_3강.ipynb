## 3-2 예외 처리

예외처리 중요해용 데이터 분석에서는 외부 리소스 많이 쓰고 코드 길어지는데 오류 어딘지 모를 때 디버깅 오래 걸리니까 분석 시에는 예외 처리 적절히 해줘야함
- 오류 별도 처리 안하면 프로그램 즉시 종료
- 사용자 경험 저하, 데이터 손실, 리소스 누수, 디버깅 어려움 등 추가적 문제 발생

ㅇ 예외 및 예외 처리
- 예외(exception): 예기치 않은 오류
- 예외처리(exception handling): 프로그램의 비정상적 종료 방지
- try-except문과 finally 블록을 사용

ㅇ try-except문
number = int(input("숫자를 입력하세요: "))
result = 10/number
언뜻 보면 오류 발생 안할거같은데 사용자가 문자 입력한다 또는 0을 입력한다 이러면 오류남

try:
    number = int(input("숫자를 입력하세요: "))
    result = 10/number
except ZeroDivisionError:
    print("0으로 나눌 수 없습니다.")
0으로 나눌 수 없습니다 except로 ZeroDivisionError 캐치해서 써줘야함

또 다른 오류 방지하기 위해 밑에 또 써줄 수 있음
얘는 int아닌데 int로 처리하면 나타나는 오류 방지용
except ValueError:
    print("유효한 숫자를 입력해야합니다.")

ㅇ finally 블록의 사용
- 예외 발생 여부와 관계없ㅇ 반드시 실행되는 코드 블록

try:
    file = open("data.csv", "r")
    content = file.read()
    print(content)
except FileNotFoundError:
    print("파일을 찾을 수 없습니다.")

운영체제에 파일이 없을 때 오류 방지용으로 넣어줫서여 근데 뭐가 문젤까요?
리소스 사용하기 위해선 허가-사용-폐기 해야되는데 지금 페기 없음
오류 발생해도 파일포인터 받아왔기 때문에 꼭 폐기해줘야함

try:
    file = open("data.csv", "r")
    content = file.read()
    print(content)
except FileNotFoundError:
    print("파일을 찾을 수 없습니다.")
finally:
    print("파일 작업 종료.")
    file.close()

앞에서 오류나면 컨텍스트 관리안됨
근데 만약 except 다음 종료 넣으면 정상적 실행 시 폐기 안됨
그래서 정상 실행되어도, exception이 발생해도 finally로 반드시 닫히도록함

## 3-3 함수형 프로그래밍

사실은 좀 큰 개념임 파이썬은 절차적 프로그래밍, 객체지향 프로그래밍, 함수형 프로그래밍 방식 지원

함수형 프로그래밍: 데이터 변경을 최소화하고 순수 함수(pure fuction)를 활용하여 부작용(side effect)을 줄이는 것을 목표로함
- 순수함수: 같은 입력에 대해 항상 같은 출력을 반환하는 함수
- 부작용: 함수가 단순히 값을 계산하는 것 외에 외부에 영향을 미치는 모든 행위
코드 가독성 향상, 디버깅 용이, 병렬처리 효율 수행

시점, 파일위치 등에 따라 막 다른 값이 출력될 수도 있음
변수에 값을 할당하지 않는게 함수형 프로그래밍
재귀적 함수만 써서 문제를 해결하는게 함수형 프로그래밍이래;;

1. 람다 함수
 - 이름이 없는 익명 함수(anonymous function)
 - 한 줄의 표현식으로 작성할 수 있는 간결한 함수
 - 즉시 사용할 수 있는 간결한 연산을 수행할 때 적합
 - 단 하나의 표현식만 허용, 표현식의 결과는 자동으로 함수의 반환값이 되므로 return 키워드 사용햐지 않음

 lambda 매개변수: 표현식

* 일반 함수
def add(x):
    return x + 2
print(add(3))

* 람다 함수
add_lambda = lambda x: x + 2
print(add_lambda(3))

2. map, filter, reduce 함수
- 대표적인 함수형 프로그래밍 기법을 사용하는 함수
- 반복문을 사용하지 않고 리스트나 반복 가능한 객체의 요소를 손쉽게 조작할 수 있도록 지원
- 데이터 변환, 필터링, 누적 연산을 수행

1) map: 통과일 > 까서 슬라이스 처럼 로직을 각각의 요소에 적용해서 변환된 형태로 만들어줌
 - 반복 가능한 객체의 모든 요소에 특정 함수를 적용하여 새로운 값을 생성하는 함수
 - 리스트나 튜플과 같은 자료구조의 요소를 변경할 때 유용

2) filter: 사과, 오렌지, 바나나 > 오렌지, 어떤 특성만 만족하는거 걸러줌
 - 반복 가능한 객체의 요소 중 특정 조건을 만족하는 요소만 남기는 함수
 - True 또는 False를 반환하는 함수(조건 함수)를 인자로 받고 조건 함수가 True를 반환하는 요소만 컬렉션에 포함

3) reduce: 사과, 오렌지, 바나나 > 과일샐러드, 조작해서 하나의 값으로 만듦
